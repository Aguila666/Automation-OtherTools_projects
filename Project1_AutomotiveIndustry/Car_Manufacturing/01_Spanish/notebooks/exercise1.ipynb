## Tratando de detectar valores atípicos (outliers) o incorrectos

```python
# Seleccionar todas las columnas de tipo 'object' (normalmente cadenas de texto)
text_columns = df.select_dtypes(include='object').columns

# Mostrar las columnas de texto
print(text_columns)
```
