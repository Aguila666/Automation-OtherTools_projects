### 1. Cargar dataset limpio

```python
import pandas as pd
import seaborn as sns
```

```python
file_path = '../datasets/processed/car_manufacturing_clean.csv'
```

```python
df = pd.read_csv(file_path)
```

### 2. Análisis exploratorio de datos

```python
# 2.1 Revisión de distribución
df['Cost'].hist()
```

```python
# 2.2 Revisar visualmente relaciones

sns.scatterplot(x='Defects', y='Cost', data=df)
```

```python
# 2.3 Ver correlación con Cost

df.corr(numeric_only=True)['Cost'].sort_values(ascending=False)
```

```python
# 2.4 Correlación de todas las features con Cost (get_dummies)

df_ml = pd.get_dummies(
    df,
    columns=['Engine_Type', 'Production_Line'],
    drop_first=True
)

df_ml.corr(numeric_only=True)['Cost'].sort_values(ascending=False)
```
